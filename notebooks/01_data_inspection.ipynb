```python
import pandas as pd
import numpy as np

# ==========================================
# 1. Load cleaned dataset
# ==========================================

clean_df = pd.read_csv(
    "../data/processed/swiggy_cleaned.csv"
)

print("Initial shape:", clean_df.shape)


# ==========================================
# 2. Clean geographic coordinates
# ==========================================

coordinate_columns = [
    "Restaurant_latitude",
    "Restaurant_longitude",
    "Delivery_location_latitude",
    "Delivery_location_longitude"
]

for column in coordinate_columns:
    clean_df[column] = clean_df[column].abs()


# ==========================================
# 3. Calculate Haversine distance
# ==========================================

def haversine_distance(lat1, lon1, lat2, lon2):

    R = 6371

    lat1 = np.radians(lat1)
    lat2 = np.radians(lat2)

    dlat = lat2 - lat1
    dlon = np.radians(lon2) - np.radians(lon1)

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arctan2(
        np.sqrt(a),
        np.sqrt(1 - a)
    )

    return R * c


clean_df["Distance_km"] = haversine_distance(
    clean_df["Restaurant_latitude"],
    clean_df["Restaurant_longitude"],
    clean_df["Delivery_location_latitude"],
    clean_df["Delivery_location_longitude"]
)


# ==========================================
# 4. Convert Order_Date to datetime
# ==========================================

clean_df["Order_Date"] = pd.to_datetime(
    clean_df["Order_Date"],
    errors="coerce"
)


# ==========================================
# 5. Create time features
# ==========================================

order_time = pd.to_datetime(
    clean_df["Time_Orderd"],
    format="%H:%M:%S",
    errors="coerce"
)

clean_df["order_hour"] = order_time.dt.hour

clean_df["day_of_week"] = (
    clean_df["Order_Date"].dt.dayofweek
)

clean_df["month"] = (
    clean_df["Order_Date"].dt.month
)

clean_df["is_weekend"] = (
    clean_df["day_of_week"] >= 5
).astype(int)


# ==========================================
# 6. Create pickup delay
# ==========================================

pickup_time = pd.to_datetime(
    clean_df["Time_Order_picked"],
    format="%H:%M:%S",
    errors="coerce"
)

order_minutes = (
    order_time.dt.hour * 60
    + order_time.dt.minute
)

pickup_minutes = (
    pickup_time.dt.hour * 60
    + pickup_time.dt.minute
)

pickup_delay = (
    pickup_minutes - order_minutes
)

# Handle midnight crossover
pickup_delay = pickup_delay.where(
    pickup_delay >= 0,
    pickup_delay + 1440
)

clean_df["pickup_delay_minutes"] = pickup_delay


# ==========================================
# 7. Create peak-hour feature
# ==========================================

clean_df["is_peak_hour"] = (
    clean_df["order_hour"].between(12, 14)
    |
    clean_df["order_hour"].between(19, 22)
).astype(int)


# ==========================================
# 8. Show resulting columns
# ==========================================

print("\nFeature engineering completed!")

print("\nNew features:")
print([
    "Distance_km",
    "order_hour",
    "day_of_week",
    "month",
    "is_weekend",
    "pickup_delay_minutes",
    "is_peak_hour"
])

print("\nFinal shape:", clean_df.shape)
```
